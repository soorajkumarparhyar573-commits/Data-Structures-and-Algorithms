## Create the Tree

```python
class Tree:
    class Position:
        def __init__(self, node):
            self._node = node

        def element(self):
            return self._node.data

    class Node:
        def __init__(self, data, parent=None):
            self.data = data
            self.parent = parent
            self.children = []

    def __init__(self):
        self.root = None

    def _add_root(self, data):
        self.root = self.Node(data)
        return self.Position(self.root)

    def _add_child(self, p, data):
        parent = p._node
        child = self.Node(data, parent)
        parent.children.append(child)
        return self.Position(child)

    def is_leaf(self, p):
        return len(p._node.children) == 0

    def positions(self):
        result = []

        def traverse(node):
            if node is None:
                return
            result.append(self.Position(node))
            for child in node.children:
                traverse(child)

        traverse(self.root)
        return result

    def depth(self, p):
        if p._node.parent is None:
            return 0
        return 1 + self.depth(self.Position(p._node.parent))

    def height(self):
        def calculate(node):
            if len(node.children) == 0:
                return 0
            return 1 + max(calculate(c) for c in node.children)

        return calculate(self.root)

    def __len__(self):
        return len(self.positions())


# Create Tree
T = Tree()

root = T._add_root('A')

b = T._add_child(root, 'B')
c = T._add_child(root, 'C')
d = T._add_child(root, 'D')

e = T._add_child(b, 'E')
f = T._add_child(b, 'F')
g = T._add_child(b, 'G')

h = T._add_child(d, 'H')

i = T._add_child(g, 'I')

print("Tree Created Successfully")

for p in T.positions():
    print(p.element())
```

## Find the Height of the Tree

```python
print("Height =", T.height())
```

 ## Find Depth of Nodes

```python
print("Depth of A =", T.depth(root))
print("Depth of B =", T.depth(b))
print("Depth of G =", T.depth(g))
print("Depth of I =", T.depth(i))
```

## Identify Internal Nodes

```python
print("Internal Nodes:")

for p in T.positions():
    if not T.is_leaf(p):
        print(p.element())
```

 ## Identify External Nodes

```python
print("External Nodes:")

for p in T.positions():
    if T.is_leaf(p):
        print(p.element())
```

 ## Calculate Tree Size

```python
print("Size =", len(T))
```

## Count Internal and External Nodes

```python
internal = 0
external = 0

for p in T.positions():

    if T.is_leaf(p):
        external += 1
    else:
        internal += 1

print("Internal Nodes =", internal)
print("External Nodes =", external)
```

 ## Add New Nodes

```python
j = T._add_child(c, 'J')
k = T._add_child(j, 'K')
l = T._add_child(k, 'L')

print("New Height =", T.height())
print("New Size =", len(T))

print("Internal Nodes:")

for p in T.positions():
    if not T.is_leaf(p):
        print(p.element())
```

```python
def count_descendants(T, p):

    count = 0

    for child in p._node.children:
        count += 1
        count += count_descendants(T, T.Position(child))

    return count


print("Descendants of B =", count_descendants(T, b))
```
