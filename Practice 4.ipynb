```python
from collections import deque

queue = deque()
# Add all tokens
for token in [101, 102, 103, 104, 105]:
    queue.append(token)
print("Initial Queue:", list(queue))

# Serve three students
for i in range(3):
    print("Served:", queue.popleft())

print("Queue after serving 3:", list(queue))
# Add tokens 106 and 107
queue.append(106)
queue.append(107)
print("Queue after adding 106 and 107:", list(queue))
# Continue serving
while queue:
    print("Served:", queue.popleft())
```

```python
import heapq

priority_queue = []

students = [
    ("Ahmed", 85),
    ("Sara", 93),
    ("Bilal", 77),
    ("Maria", 97),
    ("Hamza", 88)
]
# Add students
for name, merit in students:
    heapq.heappush(priority_queue, (-merit, name))

print("Removal Order:")

# Remove students according to priority
while priority_queue:
    merit, name = heapq.heappop(priority_queue)
    print(name, "Merit:", -merit)
```

```python
from collections import deque

dq = deque()
# Step 1
dq.appendleft("A")
print("1. insert_first(A):", list(dq))
# Step 2
dq.append("B")
print("2. insert_last(B):", list(dq))

# Step 3
dq.appendleft("C")
print("3. insert_first(C):", list(dq))

# Step 4
dq.append("D")
print("4. insert_last(D):", list(dq))
# Step 5
dq.popleft()
print("5. delete_first():", list(dq))
# Step 6
dq.pop()
print("6. delete_last():", list(dq))
```

```python
from collections import deque

history = deque()
# Visit pages
websites = [
    "google.com",
    "duet.edu.pk",
    "wiley.com",
    "python.org",
    "github.com"]
for website in websites:
    history.append(website)

print("Browser History:")
print(list(history))
# Move backward
print("\nBackward Navigation:")
while history:
    print("Back:", history.pop())
```

```python
from collections import deque
import heapq

data = [10, 20, 30, 40, 50]
# Queue
queue = deque(data)
print("Queue:")
while queue:
    print(queue.popleft(), end=" ")
# Priority Queue
priority_queue = []
for item in data:
    heapq.heappush(priority_queue, item)
print("\n\nPriority Queue:")
while priority_queue:
    print(heapq.heappop(priority_queue), end=" ")
# Deque
dq = deque(data)
print("\n\nDeque:")
print("Front:", dq[0])
print("Rear:", dq[-1])
dq.appendleft(5)
dq.append(60)
print("After insertion:", list(dq))
dq.popleft()
dq.pop()

print("After deletion:", list(dq))
```
