## Experimental Study of Time Complexity for List Operations

```python
import time
my_list = [1, 2, 3, 4, 5]
print("--- Testing for List Size: 5 ---")
# Experiment 1: Traversing and printing all elements O(n)
start_time_n = time.time()
for item in my_list:
    print(item, end=" ")
end_time_n = time.time()
time_O_n = end_time_n - start_time_n
print("\n")
start_time_1 = time.time()
first_element = my_list[0]
end_time_1 = time.time()
time_O_1 = end_time_1 - start_time_1
print("-" * 50)
print(f"Time to Print Entire List: {time_O_n:.6f} seconds")
print(f"Time to Access First Element: {time_O_1:.6f} seconds")
print("-" * 50)
```

```python
def maximum(x):
    m = x[0]
    for i in x:
        if i > m:
            m = i
    return m

numbers = [34, 32, 4, 76, 98]
print("Maximum =", maximum(numbers))
```

```python
def sort_list(x):
    for b in range(len(x)):
        for a in range(b + 1, len(x)):
            if x[b] > x[a]:
                x[b], x[a] = x[a], x[b]
    return x

numbers = [5, 3, 8, 2, 3]
print("Sorted list =", sort_list(numbers))
```
