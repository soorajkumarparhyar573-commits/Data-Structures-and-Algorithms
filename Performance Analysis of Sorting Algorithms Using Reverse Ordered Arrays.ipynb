```python
def bubble_sort(arr):
    n = len(arr)

    for i in range(n):
        for j in range(0, n - i - 1):

            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

    return arr


# Example
arr = [5, 4, 3, 2, 1]

print("Before Sorting:", arr)
bubble_sort(arr)
print("After Sorting:", arr)
```

```python
def selection_sort(arr):
    n = len(arr)

    for i in range(n):

        min_index = i

        for j in range(i + 1, n):

            if arr[j] < arr[min_index]:
                min_index = j

        arr[i], arr[min_index] = arr[min_index], arr[i]

    return arr


# Example
arr = [5, 4, 3, 2, 1]

print("Before Sorting:", arr)
selection_sort(arr)
print("After Sorting:", arr)
```

```python
def insertion_sort(arr):

    for i in range(1, len(arr)):

        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:

            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

    return arr


# Example
arr = [5, 4, 3, 2, 1]

print("Before Sorting:", arr)
insertion_sort(arr)
print("After Sorting:", arr)
```

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    result = []

    i = 0
    j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result
# Example
arr = [5, 4, 3, 2, 1]

print("Before Sorting:", arr)
print("After Sorting:", merge_sort(arr))
```

```python
def radix_sort(arr):
    maximum = max(arr)
    exp = 1
    while maximum // exp > 0:
        output = [0] * len(arr)
        count = [0] * 10
        for num in arr:
            index = (num // exp) % 10
            count[index] += 1
        for i in range(1, 10):
            count[i] += count[i - 1]
        for i in range(len(arr) - 1, -1, -1):
            index = (arr[i] // exp) % 10
            output[count[index] - 1] = arr[i]
            count[index] -= 1
        for i in range(len(arr)):
            arr[i] = output[i]
        exp *= 10
    return arr
# Example
arr = [5, 4, 3, 2, 1]
print("Before Sorting:", arr)
radix_sort(arr)
print("After Sorting:", arr)
```

```python
def bucket_sort(arr):

    if len(arr) == 0:
        return arr
    buckets = [[] for _ in range(10)]
    maximum = max(arr)
    minimum = min(arr)
    if maximum == minimum:
        return arr
    for num in arr:
        index = int(
            (num - minimum) /
            (maximum - minimum) * 9        )
        buckets[index].append(num)
    result = []
    for bucket in buckets:
        bucket.sort()
        result.extend(bucket)
    arr[:] = result
    return arr
# Example
arr = [5, 4, 3, 2, 1]
print("Before Sorting:", arr)
bucket_sort(arr)
print("After Sorting:", arr)
```

```python
algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort,
    "Merge Sort": merge_sort,
    "Radix Sort": radix_sort,
    "Bucket Sort": bucket_sort}
sizes = [500, 5000, 10000]

for size in sizes:
    original_array = list(range(size, 0, -1))

    print("\nArray Size:", size)
    print("-" * 45)
    for name, algorithm in algorithms.items():
        arr = original_array.copy()
        if name == "Merge Sort":
            sorted_array = algorithm(arr)
        else:
            algorithm(arr)
            sorted_array = arr
        print(name, ":", sorted_array[:10])
```

```python
import time

algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort,
    "Merge Sort": merge_sort,
    "Radix Sort": radix_sort,
    "Bucket Sort": bucket_sort}
sizes = [500, 5000, 10000]
results = {}
for size in sizes:
    original_array = list(range(size, 0, -1))
    print("\nArray Size:", size)
    print("-" * 45)
    results[size] = {}
    for name, algorithm in algorithms.items():
        arr = original_array.copy()
        start_time = time.perf_counter()
        algorithm(arr)
        end_time = time.perf_counter()
        # Convert seconds to milliseconds
        execution_time = (end_time - start_time) * 1000
        results[size][name] = execution_time
        print(
            name,
            ":",
            round(execution_time, 4),
            "ms")
```
