```python
import time

# Function to measure push operation time
def measure_push(n):
    stack = []
    start_time = time.perf_counter()

    for i in range(n):
        stack.append(i)   # Push operation
    end_time = time.perf_counter()

    return end_time - start_time

sizes = [10000, 100000, 500000, 1000000]

print("Number of Elements\tPush Time (seconds)")

for n in sizes:
    time_taken = measure_push(n)
    print(n, "\t\t\t", time_taken)
```

```python
import time

sizes = [10000, 100000, 500000, 1000000]

for n in sizes:
    # Create and fill the stack
    stack = list(range(n))

    # Start timer
    start_time = time.perf_counter()

    # Pop all elements
    while stack:
        stack.pop()

    # Stop timer
    end_time = time.perf_counter()

    pop_time = end_time - start_time

    print(n, "elements -> Pop Time:", pop_time, "seconds")
```

```python
import time

stack = []

# Push 1,000,000 pages
start = time.time()

for i in range(1000000):
    stack.append("Page " + str(i))

push_time = time.time() - start

# Pop 1,000,000 pages
start = time.time()

while stack:
    stack.pop()

pop_time = time.time() - start

print("Push Time:", push_time, "seconds")
print("Pop Time:", pop_time, "seconds")
```
