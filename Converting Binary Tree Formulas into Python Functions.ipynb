```python
def node_range(h):
    minimum = h + 1
    maximum = 2 ** (h + 1) - 1
    return minimum, maximum

h = 3
minimum, maximum = node_range(h)

print("Minimum Nodes =", minimum)
print("Maximum Nodes =", maximum)
```

```python
def external_node_range(h):
    minimum = 1
    maximum = 2 ** h
    return minimum, maximum

h = 3
minimum, maximum = external_node_range(h)

print("Minimum External Nodes =", minimum)
print("Maximum External Nodes =", maximum)
```

```python
def internal_node_range(h):
    minimum = h
    maximum = 2 ** h - 1
    return minimum, maximum

h = 3
minimum, maximum = internal_node_range(h)

print("Minimum Internal Nodes =", minimum)
print("Maximum Internal Nodes =", maximum)
```

```python
import math

def height_range(n):
    minimum = math.ceil(math.log2(n + 1)) - 1
    maximum = n - 1
    return minimum, maximum

n = 15
minimum, maximum = height_range(n)

print("Minimum Height =", minimum)
print("Maximum Height =", maximum)
```

```python
def node_range(h):
    return h + 1, (2 ** (h + 1)) - 1

def external_node_range(h):
    return 1, 2 ** h
def internal_node_range(h):
    return h, (2 ** h) - 1

def height_range(n):
    return n.bit_length() - 1, n - 1

while True:
    print("\n===== BINARY TREE PROPERTIES =====")
    print("1. Find Range of Total Nodes")
    print("2. Find Range of External Nodes")
    print("3. Find Range of Internal Nodes")
    print("4. Find Range of Height")
    print("5. Exit")

    choice = int(input("Enter your choice: "))
    if choice == 1:
        h = int(input("Enter tree height: "))
        if h >= 0:
            minimum, maximum = node_range(h)
            print("Minimum Nodes =", minimum)
            print("Maximum Nodes =", maximum)
        else:
            print("Invalid Height")

    elif choice == 2:
        h = int(input("Enter tree height: "))
        if h >= 0:
            minimum, maximum = external_node_range(h)
            print("Minimum External Nodes =", minimum)
            print("Maximum External Nodes =", maximum)
        else:
            print("Invalid Height")

    elif choice == 3:
        h = int(input("Enter tree height: "))
        if h >= 0:
            minimum, maximum = internal_node_range(h)
            print("Minimum Internal Nodes =", minimum)
            print("Maximum Internal Nodes =", maximum)
        else:
            print("Invalid Height")

    elif choice == 4:
        n = int(input("Enter number of nodes: "))
        if n > 0:
            minimum, maximum = height_range(n)
            print("Minimum Height =", minimum)
            print("Maximum Height =", maximum)
        else:
            print("Invalid Number of Nodes")

    elif choice == 5:
        print("Program Exited Successfully!")
        break

    else:
        print("Invalid Choice! Please Try Again.")
```
